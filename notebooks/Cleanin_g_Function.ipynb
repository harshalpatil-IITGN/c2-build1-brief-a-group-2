```python
import pandas as pd

def recover_missing_test_codes(bookings_df, results_df):
    """
    Recover missing test_code values when exactly one booked code
    is absent from the recorded results for that booking.

    Returns:
        cleaned_results: copy of results with inferred codes filled
        audit: DataFrame showing each missing row and its status
    """
    cleaned_results = results_df.copy()

    # Identify rows with missing test codes
    missing_mask = cleaned_results["test_code"].isna()

    if not missing_mask.any():
        return cleaned_results, pd.DataFrame()

    missing_ids = cleaned_results.loc[
        missing_mask, "booking_id"
    ].unique()

    # Unique booked codes for affected bookings
    booked_codes = (
        bookings_df[
            bookings_df["booking_id"].isin(missing_ids)
        ]
        .dropna(subset=["test_code"])
        .groupby("booking_id")["test_code"]
        .agg(lambda s: set(s))
    )

    # Existing recorded result codes for affected bookings
    recorded_codes = (
        cleaned_results[
            cleaned_results["booking_id"].isin(missing_ids)
        ]
        .dropna(subset=["test_code"])
        .groupby("booking_id")["test_code"]
        .agg(lambda s: set(s))
    )

    audit_rows = []

    for idx in cleaned_results.index[missing_mask]:
        booking_id = cleaned_results.at[idx, "booking_id"]

        booked = booked_codes.get(booking_id, set())
        recorded = recorded_codes.get(booking_id, set())

        candidates = booked - recorded

        if len(candidates) == 1:
            inferred_code = next(iter(candidates))
            cleaned_results.at[idx, "test_code"] = inferred_code
            status = "inferred_unique_candidate"
        else:
            inferred_code = None
            status = (
                "ambiguous"
                if len(candidates) > 1
                else "no_candidate"
            )

        audit_rows.append({
            "row_index": idx,
            "booking_id": booking_id,
            "inferred_test_code": inferred_code,
            "candidate_codes": sorted(candidates),
            "status": status
        })

    audit = pd.DataFrame(audit_rows)

    return cleaned_results, audit
```

```python

import pandas as pd


def recover_missing_test_codes(results_df, bookings_df):
    results = results_df.copy().reset_index(drop=False)
    results = results.rename(columns={"index": "_original_index"})

    bookings = bookings_df.copy()

    # Normalize IDs and codes
    for df in (results, bookings):
        df["booking_id"] = df["booking_id"].astype("string").str.strip()
        if "test_code" in df.columns:
            df["test_code"] = df["test_code"].astype("string").str.strip()
            df["test_code"] = df["test_code"].replace("", pd.NA)

    # This function expects one booked test per row
    if "test_code" not in bookings.columns:
        raise ValueError("Bookings must contain a test_code column.")

    bookings = bookings.dropna(subset=["booking_id", "test_code"])

    # Unique codes booked under each booking
    booked_by_id = (
        bookings.groupby("booking_id")["test_code"]
        .agg(lambda s: set(s.dropna()))
        .to_dict()
    )

    # Codes already recorded in results
    recorded_by_id = (
        results.dropna(subset=["booking_id", "test_code"])
        .groupby("booking_id")["test_code"]
        .agg(lambda s: set(s.dropna()))
        .to_dict()
    )

    audit_rows = []
    missing_indices = results.index[results["test_code"].isna()].tolist()

    for idx in missing_indices:
        bid = results.at[idx, "booking_id"]

        booked = booked_by_id.get(bid, set())
        recorded = recorded_by_id.get(bid, set())
        candidates = sorted(booked - recorded)

        recovered = pd.NA

        if len(candidates) == 1:
            recovered = candidates[0]
            results.at[idx, "test_code"] = recovered

            # Update immediately to prevent reusing the same code
            recorded_by_id.setdefault(bid, set()).add(recovered)
            status = "recovered_unique_candidate"

        elif len(candidates) > 1:
            status = "ambiguous_multiple_candidates"
        elif bid not in booked_by_id:
            status = "booking_not_found_or_no_codes"
        else:
            status = "no_unrecorded_candidate"

        audit_rows.append({
            "result_index": results.at[idx, "_original_index"],
            "booking_id": bid,
            "candidate_codes": candidates,
            "recovered_test_code": recovered,
            "recovery_status": status,
        })

    # Restore original result index and columns
    results = results.set_index("_original_index")
    results.index.name = results_df.index.name

    audit = pd.DataFrame(audit_rows)

    return results, audit

df_results_2026_H1_clean, audit_2026 = recover_missing_test_codes(
    df_results_2026_H1,
    df_bookings_2026_H1
)

df_results_2025_H1_clean, audit_2025 = recover_missing_test_codes(
    df_results_2025_H1,
    df_bookings_2025_H1
)

print("2026 H1")
print(audit_2026["recovery_status"].value_counts())
print("Missing codes remaining:",
      df_results_2026_H1_clean["test_code"].isna().sum())

print("\n2025 H1")
print(audit_2025["recovery_status"].value_counts())
print("Missing codes remaining:",
      df_results_2025_H1_clean["test_code"].isna().sum())
```

```python
"""
Kalpa Health, Brief A: cleaning pipeline (results tables, H1 2025 vs H1 2026)

Run in the notebook that already holds the four raw DataFrames:

    %run kalpa_clean.py
    res, book = clean_all(df_results_2025_H1, df_results_2026_H1,
                          df_bookings_H1_2025_H1, df_bookings_2026_H1)

res  = one row per test result, both years: city, parsed result_dt, test_code
       ('UNKNOWN' if it could not be recovered), flags test_code_recovered
       and test_code_unknown
book = one row per (booking, test), both years, new-format bookings exploded

The raw inputs are never modified.
"""
import ast

import pandas as pd

CITY_FIX = {"Bangalore": "Bengaluru"}
PREFIX_CITY = {
    "BK-AHM": "Ahmedabad", "BK-BLR": "Bengaluru", "BK-CHE": "Chennai",
    "BK-HYD": "Hyderabad", "BK-JAI": "Jaipur", "BK-PUN": "Pune",
    "NOVA_CHE": "Chennai", "NOVA_JAI": "Jaipur",
}
ID_PREFIX = r"^([A-Z]+[-_][A-Z]+)"


# ---------------------------------------------------------------- helpers
def _norm_id(s):
    return s.astype(str).str.strip().str.upper()


def _norm_code(s):
    return s.str.strip().str.upper()          # keeps NaN as NaN


def _to_obj(x):
    """`tests` may be a real python list or its string repr."""
    if isinstance(x, (list, dict)):
        return x
    if isinstance(x, str):
        try:
            return ast.literal_eval(x)
        except Exception:
            return None
    return None


def drop_header_rows(df, name):
    """Repeated header rows left over from concatenating the city files."""
    bad = df["booking_id"].astype(str).str.strip().str.lower().eq("booking_id")
    print(f"{name}: dropped {bad.sum()} repeated header rows")
    return df[~bad].copy()


def parse_ts(col):
    """Two formats coexist: ISO with +05:30 offset, and day-first dd/mm/yyyy HH:MM.
    Both end up as naive IST datetimes."""
    s = col.astype(str).str.strip()
    dmy = s.str.match(r"^\d{2}/\d{2}/\d{4} \d{2}:\d{2}")
    iso = s.str.match(r"^\d{4}-\d{2}-\d{2}")
    a = pd.to_datetime(s.where(dmy), format="%d/%m/%Y %H:%M", errors="coerce")
    b = (pd.to_datetime(s.where(iso), format="ISO8601", utc=True, errors="coerce")
           .dt.tz_convert("Asia/Kolkata").dt.tz_localize(None))
    out = a.fillna(b)
    print(f"  {col.name}: dd/mm/yyyy={dmy.sum():,} | ISO={iso.sum():,} | unparsed={out.isna().sum():,}")
    return out


# --------------------------------------------------------------- bookings
def _explode_new_format(b):
    """New booking software (Jaipur/Chennai): one row per booking with a `tests`
    list of {'code', 'price'} -> one row per test, like the old format."""
    is_new = b["tests"].notna()
    old, new = b[~is_new].copy(), b[is_new].copy()
    new["tests"] = new["tests"].map(_to_obj)
    new = new.explode("tests", ignore_index=True)
    new["test_code"] = new["tests"].map(lambda d: d.get("code") if isinstance(d, dict) else None)
    new["test_price"] = new["tests"].map(lambda d: d.get("price") if isinstance(d, dict) else None)
    return pd.concat([old, new], ignore_index=True)


def prep_bookings(raw, year):
    b = drop_header_rows(raw, f"bookings {year}")
    if "tests" in b.columns:
        b = _explode_new_format(b)
    if "rescheduled_from" not in b.columns:
        b["rescheduled_from"] = pd.Series([None] * len(b), index=b.index, dtype="object")
    return pd.DataFrame({
        "booking_id": _norm_id(b["booking_id"]),
        "city": b["city"].str.strip().replace(CITY_FIX),
        "test_code": _norm_code(b["test_code"]),
        "test_price": b["test_price"],
        "created_dt": parse_ts(b["created_at"]),
        "slot_dt": parse_ts(b["slot_at"]),
        "rescheduled_from": b["rescheduled_from"].astype("object").str.strip().str.upper(),
        "year": year,
    })


# ---------------------------------------------------------------- results
def recover_test_codes(res, book):
    """Fill a null test_code from the booking: booked tests that have no coded result
    row yet are the candidates. Fill only when exactly 1 null row and exactly 1
    candidate exist for that booking, so a fill can never create a duplicate pair."""
    booked = book.dropna(subset=["test_code"])[["booking_id", "test_code"]].drop_duplicates()
    done = res.dropna(subset=["test_code"])[["booking_id", "test_code"]]
    m = booked.merge(done, on=["booking_id", "test_code"], how="left", indicator=True)
    cand = (m[m["_merge"] == "left_only"]
            .groupby("booking_id")["test_code"].agg(list).rename("candidates"))
    n_null = res[res["test_code"].isna()].groupby("booking_id").size().rename("n_null")

    fix = pd.concat([n_null, cand], axis=1, join="inner")
    fix["n_cand"] = fix["candidates"].str.len()
    safe = fix[(fix["n_null"] == 1) & (fix["n_cand"] == 1)].copy()
    safe["fill"] = safe["candidates"].str[0]

    out = res.merge(safe["fill"], left_on="booking_id", right_index=True, how="left")
    mask = out["test_code"].isna() & out["fill"].notna()
    out.loc[mask, "test_code"] = out.loc[mask, "fill"]
    out["test_code_recovered"] = mask
    print(f"  null test_code: {len(res) - len(res.dropna(subset=['test_code'])):,} | "
          f"recovered: {int(mask.sum()):,} | still null: {int(out['test_code'].isna().sum()):,}")
    return out.drop(columns="fill")


def prep_results(raw, book, year):
    r = drop_header_rows(raw, f"results {year}")
    r["booking_id"] = _norm_id(r["booking_id"])
    r["test_code"] = _norm_code(r["test_code"])
    r = recover_test_codes(r, book)
    r["test_code_unknown"] = r["test_code"].isna()
    r["test_code"] = r["test_code"].fillna("UNKNOWN")

    # city: from the booking if it exists, otherwise from the booking_id prefix
    r["prefix"] = r["booking_id"].str.extract(ID_PREFIX)[0]
    city_map = book.drop_duplicates("booking_id").set_index("booking_id")["city"]
    r["city"] = r["booking_id"].map(city_map).fillna(r["prefix"].map(PREFIX_CITY))

    r["result_dt"] = parse_ts(r["result_ts"])
    r["year"] = year
    return r


# --------------------------------------------------------------------- QA
def qa_report(res, book):
    print("\n================ QA REPORT ================")
    print(res.groupby("year").agg(
        rows=("booking_id", "size"),
        recovered=("test_code_recovered", "sum"),
        unknown=("test_code_unknown", "sum"),
        no_city=("city", lambda s: int(s.isna().sum()))))

    print("\nresult months (expect mostly 1-6):")
    print(res.groupby(["year", res["result_dt"].dt.month]).size().unstack(0))

    ct = res.groupby(["city", "year"]).size().unstack()
    ct.loc["TOTAL"] = ct.sum()
    ct["growth_pct"] = (ct[2026] / ct[2025] - 1) * 100
    print("\nresults by city:")
    print(ct.round(1))

    pk = res["test_code"].str.startswith("PK").rename("is_package")
    pt = res.groupby([pk, "year"]).size().unstack()
    pt["growth_pct"] = (pt[2026] / pt[2025] - 1) * 100
    print("\npackages vs other tests:")
    print(pt.round(1))

    # bookings side; rescheduled bookings get a new id with no results, so remove them
    resched = book.loc[book["rescheduled_from"].notna(), "booking_id"].unique()
    book_adj = book[~book["booking_id"].isin(resched)]
    bk = book.groupby(["city", "year"]).size().unstack()
    bka = book_adj.groupby(["city", "year"]).size().unstack()
    rr = res.groupby(["city", "year"]).size().unstack()
    chk = pd.DataFrame({
        "booked_growth": (bk[2026] / bk[2025] - 1) * 100,
        "booked_excl_resched_growth": (bka[2026] / bka[2025] - 1) * 100,
        "results_growth": (rr[2026] / rr[2025] - 1) * 100,
        "results_per_booked_2026": rr[2026] / bka[2026],
    })
    print("\nbookings vs results (rescheduled duplicates removed from bookings):")
    print(chk.round(3))


# ------------------------------------------------------------------- main
def clean_all(results_2025, results_2026, bookings_2025, bookings_2026):
    book25 = prep_bookings(bookings_2025, 2025)
    book26 = prep_bookings(bookings_2026, 2026)
    res25 = prep_results(results_2025, book25, 2025)
    res26 = prep_results(results_2026, book26, 2026)
    res = pd.concat([res25, res26], ignore_index=True)
    book = pd.concat([book25, book26], ignore_index=True)
    qa_report(res, book)
    return res, book
```

```python
def growth_tree(b, r, by):
    d = pd.concat({
        "bookings": b.groupby([by, "year"])["booking_id"].nunique(),
        "booked_rows": b.groupby([by, "year"]).size(),
        "results": r.groupby([by, "year"]).size(),
    }, axis=1).unstack("year")
    g = pd.DataFrame({
        "bookings": d["bookings"][2026] / d["bookings"][2025],
        "tests_per_booking": (d["booked_rows"][2026] / d["bookings"][2026])
                             / (d["booked_rows"][2025] / d["bookings"][2025]),
        "results_per_test_booked": (d["results"][2026] / d["booked_rows"][2026])
                                   / (d["results"][2025] / d["booked_rows"][2025]),
    })
    g["product"] = g.prod(axis=1)
    g["actual"] = d["results"][2026] / d["results"][2025]
    out = ((g - 1) * 100).round(2)                      # growth in % per branch
    out["gap_pp"] = (100 * (g["product"] - g["actual"])).round(4)   # must be 0
    return out

resched = book.loc[book["rescheduled_from"].notna(), "booking_id"].unique()
b = book[~book["booking_id"].isin(resched)]

print(growth_tree(b.assign(all="ALL"), res.assign(all="ALL"), "all"))   # overall
print(growth_tree(b, res, "city"))                                      # per city
```

```python
"""
Kalpa Health, Brief A: cleaning pipeline (results tables, H1 2025 vs H1 2026)

Run in the notebook that already holds the four raw DataFrames:

    %run kalpa_clean.py
    res, book = clean_all(df_results_2025_H1, df_results_2026_H1,
                          df_bookings_H1_2025_H1, df_bookings_2026_H1)

res  = one row per test result, both years: city, parsed result_dt, test_code
       ('UNKNOWN' if it could not be recovered), flags test_code_recovered
       and test_code_unknown
book = one row per (booking, test), both years, new-format bookings exploded

The raw inputs are never modified.
"""
import ast

import pandas as pd

CITY_FIX = {"Bangalore": "Bengaluru"}
PREFIX_CITY = {
    "BK-AHM": "Ahmedabad", "BK-BLR": "Bengaluru", "BK-CHE": "Chennai",
    "BK-HYD": "Hyderabad", "BK-JAI": "Jaipur", "BK-PUN": "Pune",
    "NOVA_CHE": "Chennai", "NOVA_JAI": "Jaipur",
}
ID_PREFIX = r"^([A-Z]+[-_][A-Z]+)"


# ---------------------------------------------------------------- helpers
def _norm_id(s):
    return s.astype(str).str.strip().str.upper()


def _norm_code(s):
    return s.str.strip().str.upper()          # keeps NaN as NaN


def _to_obj(x):
    """`tests` may be a real python list or its string repr."""
    if isinstance(x, (list, dict)):
        return x
    if isinstance(x, str):
        try:
            return ast.literal_eval(x)
        except Exception:
            return None
    return None


def drop_header_rows(df, name):
    """Repeated header rows left over from concatenating the city files."""
    bad = df["booking_id"].astype(str).str.strip().str.lower().eq("booking_id")
    print(f"{name}: dropped {bad.sum()} repeated header rows")
    return df[~bad].copy()


def parse_ts(col):
    """Two formats coexist: ISO with +05:30 offset, and day-first dd/mm/yyyy HH:MM.
    Both end up as naive IST datetimes."""
    s = col.astype(str).str.strip()
    dmy = s.str.match(r"^\d{2}/\d{2}/\d{4} \d{2}:\d{2}")
    iso = s.str.match(r"^\d{4}-\d{2}-\d{2}")
    a = pd.to_datetime(s.where(dmy), format="%d/%m/%Y %H:%M", errors="coerce")
    b = (pd.to_datetime(s.where(iso), format="ISO8601", utc=True, errors="coerce")
           .dt.tz_convert("Asia/Kolkata").dt.tz_localize(None))
    out = a.fillna(b)
    print(f"  {col.name}: dd/mm/yyyy={dmy.sum():,} | ISO={iso.sum():,} | unparsed={out.isna().sum():,}")
    return out


# --------------------------------------------------------------- bookings
def _explode_new_format(b):
    """New booking software (Jaipur/Chennai): one row per booking with a `tests`
    list of {'code', 'price'} -> one row per test, like the old format."""
    is_new = b["tests"].notna()
    old, new = b[~is_new].copy(), b[is_new].copy()
    new["tests"] = new["tests"].map(_to_obj)
    new = new.explode("tests", ignore_index=True)
    new["test_code"] = new["tests"].map(lambda d: d.get("code") if isinstance(d, dict) else None)
    new["test_price"] = new["tests"].map(lambda d: d.get("price") if isinstance(d, dict) else None)
    # the new software nests the patient: {'patient_id': ..., 'is_new': ...}
    pat = new["patient"].map(_to_obj)
    new["patient_id"] = pat.map(lambda d: d.get("patient_id") if isinstance(d, dict) else None)
    new["is_new_patient"] = pat.map(lambda d: d.get("is_new") if isinstance(d, dict) else None)
    return pd.concat([old, new], ignore_index=True)


def prep_bookings(raw, year):
    b = drop_header_rows(raw, f"bookings {year}")
    if "tests" in b.columns:
        b = _explode_new_format(b)
    if "rescheduled_from" not in b.columns:
        b["rescheduled_from"] = pd.Series([None] * len(b), index=b.index, dtype="object")
    return pd.DataFrame({
        "booking_id": _norm_id(b["booking_id"]),
        "patient_id": _norm_code(b["patient_id"]),
        "is_new_patient": b["is_new_patient"] if "is_new_patient" in b.columns else None,
        "city": b["city"].str.strip().replace(CITY_FIX),
        "test_code": _norm_code(b["test_code"]),
        "test_price": b["test_price"],
        "created_dt": parse_ts(b["created_at"]),
        "slot_dt": parse_ts(b["slot_at"]),
        "rescheduled_from": b["rescheduled_from"].astype("object").str.strip().str.upper(),
        "year": year,
    })


# ---------------------------------------------------------------- results
def recover_test_codes(res, book):
    """Fill a null test_code from the booking: booked tests that have no coded result
    row yet are the candidates. Fill only when exactly 1 null row and exactly 1
    candidate exist for that booking, so a fill can never create a duplicate pair."""
    booked = book.dropna(subset=["test_code"])[["booking_id", "test_code"]].drop_duplicates()
    done = res.dropna(subset=["test_code"])[["booking_id", "test_code"]]
    m = booked.merge(done, on=["booking_id", "test_code"], how="left", indicator=True)
    cand = (m[m["_merge"] == "left_only"]
            .groupby("booking_id")["test_code"].agg(list).rename("candidates"))
    n_null = res[res["test_code"].isna()].groupby("booking_id").size().rename("n_null")

    fix = pd.concat([n_null, cand], axis=1, join="inner")
    fix["n_cand"] = fix["candidates"].str.len()
    safe = fix[(fix["n_null"] == 1) & (fix["n_cand"] == 1)].copy()
    safe["fill"] = safe["candidates"].str[0]

    out = res.merge(safe["fill"], left_on="booking_id", right_index=True, how="left")
    mask = out["test_code"].isna() & out["fill"].notna()
    out.loc[mask, "test_code"] = out.loc[mask, "fill"]
    out["test_code_recovered"] = mask
    print(f"  null test_code: {len(res) - len(res.dropna(subset=['test_code'])):,} | "
          f"recovered: {int(mask.sum()):,} | still null: {int(out['test_code'].isna().sum()):,}")
    return out.drop(columns="fill")


def prep_results(raw, book, year):
    r = drop_header_rows(raw, f"results {year}")
    r["booking_id"] = _norm_id(r["booking_id"])
    r["test_code"] = _norm_code(r["test_code"])
    r = recover_test_codes(r, book)
    r["test_code_unknown"] = r["test_code"].isna()
    r["test_code"] = r["test_code"].fillna("UNKNOWN")

    # city: from the booking if it exists, otherwise from the booking_id prefix
    r["prefix"] = r["booking_id"].str.extract(ID_PREFIX)[0]
    city_map = book.drop_duplicates("booking_id").set_index("booking_id")["city"]
    r["city"] = r["booking_id"].map(city_map).fillna(r["prefix"].map(PREFIX_CITY))

    r["result_dt"] = parse_ts(r["result_ts"])
    r["year"] = year
    return r


# --------------------------------------------------------------------- QA
def qa_report(res, book):
    print("\n================ QA REPORT ================")
    print(res.groupby("year").agg(
        rows=("booking_id", "size"),
        recovered=("test_code_recovered", "sum"),
        unknown=("test_code_unknown", "sum"),
        no_city=("city", lambda s: int(s.isna().sum()))))

    print("\nbookings rows with no patient_id (expect 0):", int(book["patient_id"].isna().sum()))

    print("\nresult months (expect mostly 1-6):")
    print(res.groupby(["year", res["result_dt"].dt.month]).size().unstack(0))

    ct = res.groupby(["city", "year"]).size().unstack()
    ct.loc["TOTAL"] = ct.sum()
    ct["growth_pct"] = (ct[2026] / ct[2025] - 1) * 100
    print("\nresults by city:")
    print(ct.round(1))

    pk = res["test_code"].str.startswith("PK").rename("is_package")
    pt = res.groupby([pk, "year"]).size().unstack()
    pt["growth_pct"] = (pt[2026] / pt[2025] - 1) * 100
    print("\npackages vs other tests:")
    print(pt.round(1))

    # bookings side; rescheduled bookings get a new id with no results, so remove them
    resched = book.loc[book["rescheduled_from"].notna(), "booking_id"].unique()
    book_adj = book[~book["booking_id"].isin(resched)]
    bk = book.groupby(["city", "year"]).size().unstack()
    bka = book_adj.groupby(["city", "year"]).size().unstack()
    rr = res.groupby(["city", "year"]).size().unstack()
    chk = pd.DataFrame({
        "booked_growth": (bk[2026] / bk[2025] - 1) * 100,
        "booked_excl_resched_growth": (bka[2026] / bka[2025] - 1) * 100,
        "results_growth": (rr[2026] / rr[2025] - 1) * 100,
        "results_per_booked_2026": rr[2026] / bka[2026],
    })
    print("\nbookings vs results (rescheduled duplicates removed from bookings):")
    print(chk.round(3))


# ------------------------------------------------------------------- main
def clean_all(results_2025, results_2026, bookings_2025, bookings_2026):
    book25 = prep_bookings(bookings_2025, 2025)
    book26 = prep_bookings(bookings_2026, 2026)
    res25 = prep_results(results_2025, book25, 2025)
    res26 = prep_results(results_2026, book26, 2026)
    res = pd.concat([res25, res26], ignore_index=True)
    book = pd.concat([book25, book26], ignore_index=True)
    qa_report(res, book)
    return res, book
```

```python
"""
Kalpa Health, Brief A: cleaning pipeline (results tables, H1 2025 vs H1 2026)

Run in the notebook that already holds the four raw DataFrames:

    %run kalpa_clean.py
    res, book = clean_all(df_results_2025_H1, df_results_2026_H1,
                          df_bookings_H1_2025_H1, df_bookings_2026_H1)

res  = one row per test result, both years: city, parsed result_dt, test_code
       ('UNKNOWN' if it could not be recovered), flags test_code_recovered
       and test_code_unknown
book = one row per (booking, test), both years, new-format bookings exploded

The raw inputs are never modified.
"""
import ast

import pandas as pd

CITY_FIX = {"Bangalore": "Bengaluru"}
PREFIX_CITY = {
    "BK-AHM": "Ahmedabad", "BK-BLR": "Bengaluru", "BK-CHE": "Chennai",
    "BK-HYD": "Hyderabad", "BK-JAI": "Jaipur", "BK-PUN": "Pune",
    "NOVA_CHE": "Chennai", "NOVA_JAI": "Jaipur",
}
ID_PREFIX = r"^([A-Z]+[-_][A-Z]+)"


# ---------------------------------------------------------------- helpers
def _norm_id(s):
    return s.astype(str).str.strip().str.upper()


def _norm_code(s):
    return s.str.strip().str.upper()          # keeps NaN as NaN


def _to_obj(x):
    """`tests` may be a real python list or its string repr."""
    if isinstance(x, (list, dict)):
        return x
    if isinstance(x, str):
        try:
            return ast.literal_eval(x)
        except Exception:
            return None
    return None


def drop_header_rows(df, name):
    """Repeated header rows left over from concatenating the city files."""
    bad = df["booking_id"].astype(str).str.strip().str.lower().eq("booking_id")
    print(f"{name}: dropped {bad.sum()} repeated header rows")
    return df[~bad].copy()


def parse_ts(col):
    """Two formats coexist: ISO with +05:30 offset, and day-first dd/mm/yyyy HH:MM.
    Both end up as naive IST datetimes."""
    s = col.astype(str).str.strip()
    dmy = s.str.match(r"^\d{2}/\d{2}/\d{4} \d{2}:\d{2}")
    iso = s.str.match(r"^\d{4}-\d{2}-\d{2}")
    a = pd.to_datetime(s.where(dmy), format="%d/%m/%Y %H:%M", errors="coerce")
    b = (pd.to_datetime(s.where(iso), format="ISO8601", utc=True, errors="coerce")
           .dt.tz_convert("Asia/Kolkata").dt.tz_localize(None))
    out = a.fillna(b)
    print(f"  {col.name}: dd/mm/yyyy={dmy.sum():,} | ISO={iso.sum():,} | unparsed={out.isna().sum():,}")
    return out


# --------------------------------------------------------------- bookings
def _explode_new_format(b):
    """New booking software (Jaipur/Chennai): one row per booking with a `tests`
    list of {'code', 'price'} -> one row per test, like the old format."""
    is_new = b["tests"].notna()
    old, new = b[~is_new].copy(), b[is_new].copy()
    new["tests"] = new["tests"].map(_to_obj)
    new = new.explode("tests", ignore_index=True)
    new["test_code"] = new["tests"].map(lambda d: d.get("code") if isinstance(d, dict) else None)
    new["test_price"] = new["tests"].map(lambda d: d.get("price") if isinstance(d, dict) else None)
    # the new software nests the patient: {'patient_id': ..., 'is_new': ...}
    pat = new["patient"].map(_to_obj)
    new["patient_id"] = pat.map(lambda d: d.get("patient_id") if isinstance(d, dict) else None)
    new["is_new_patient"] = pat.map(lambda d: d.get("is_new") if isinstance(d, dict) else None)
    return pd.concat([old, new], ignore_index=True)


def prep_bookings(raw, year):
    b = drop_header_rows(raw, f"bookings {year}")
    if "tests" in b.columns:
        b = _explode_new_format(b)
    if "rescheduled_from" not in b.columns:
        b["rescheduled_from"] = pd.Series([None] * len(b), index=b.index, dtype="object")
    return pd.DataFrame({
        "booking_id": _norm_id(b["booking_id"]),
        "patient_id": _norm_code(b["patient_id"]),
        "is_new_patient": b["is_new_patient"] if "is_new_patient" in b.columns else None,
        "city": b["city"].str.strip().replace(CITY_FIX),
        "channel": b["channel"].astype(str).str.strip().str.lower(),
        "test_code": _norm_code(b["test_code"]),
        "test_price": b["test_price"],
        "created_dt": parse_ts(b["created_at"]),
        "slot_dt": parse_ts(b["slot_at"]),
        "rescheduled_from": b["rescheduled_from"].astype("object").str.strip().str.upper(),
        "year": year,
    })


# ---------------------------------------------------------------- results
def recover_test_codes(res, book):
    """Fill a null test_code from the booking: booked tests that have no coded result
    row yet are the candidates. Fill only when exactly 1 null row and exactly 1
    candidate exist for that booking, so a fill can never create a duplicate pair."""
    booked = book.dropna(subset=["test_code"])[["booking_id", "test_code"]].drop_duplicates()
    done = res.dropna(subset=["test_code"])[["booking_id", "test_code"]]
    m = booked.merge(done, on=["booking_id", "test_code"], how="left", indicator=True)
    cand = (m[m["_merge"] == "left_only"]
            .groupby("booking_id")["test_code"].agg(list).rename("candidates"))
    n_null = res[res["test_code"].isna()].groupby("booking_id").size().rename("n_null")

    fix = pd.concat([n_null, cand], axis=1, join="inner")
    fix["n_cand"] = fix["candidates"].str.len()
    safe = fix[(fix["n_null"] == 1) & (fix["n_cand"] == 1)].copy()
    safe["fill"] = safe["candidates"].str[0]

    out = res.merge(safe["fill"], left_on="booking_id", right_index=True, how="left")
    mask = out["test_code"].isna() & out["fill"].notna()
    out.loc[mask, "test_code"] = out.loc[mask, "fill"]
    out["test_code_recovered"] = mask
    print(f"  null test_code: {len(res) - len(res.dropna(subset=['test_code'])):,} | "
          f"recovered: {int(mask.sum()):,} | still null: {int(out['test_code'].isna().sum()):,}")
    return out.drop(columns="fill")


def prep_results(raw, book, year):
    r = drop_header_rows(raw, f"results {year}")
    r["booking_id"] = _norm_id(r["booking_id"])
    r["test_code"] = _norm_code(r["test_code"])
    r = recover_test_codes(r, book)
    r["test_code_unknown"] = r["test_code"].isna()
    r["test_code"] = r["test_code"].fillna("UNKNOWN")

    # city: from the booking if it exists, otherwise from the booking_id prefix
    r["prefix"] = r["booking_id"].str.extract(ID_PREFIX)[0]
    city_map = book.drop_duplicates("booking_id").set_index("booking_id")["city"]
    r["city"] = r["booking_id"].map(city_map).fillna(r["prefix"].map(PREFIX_CITY))
    chan_map = book.drop_duplicates("booking_id").set_index("booking_id")["channel"]
    r["channel"] = r["booking_id"].map(chan_map)          # NaN = result has no booking

    r["result_dt"] = parse_ts(r["result_ts"])
    r["year"] = year
    return r


# --------------------------------------------------------------------- QA
def qa_report(res, book):
    print("\n================ QA REPORT ================")
    print(res.groupby("year").agg(
        rows=("booking_id", "size"),
        recovered=("test_code_recovered", "sum"),
        unknown=("test_code_unknown", "sum"),
        no_city=("city", lambda s: int(s.isna().sum()))))

    print("\nbookings rows with no patient_id (expect 0):", int(book["patient_id"].isna().sum()))

    print("\nchannel values (check for spelling variants between years):")
    print(book.drop_duplicates("booking_id").groupby(["channel", "year"]).size().unstack(fill_value=0))
    print("results with no booking (no channel):", int(res["channel"].isna().sum()))

    print("\nresult months (expect mostly 1-6):")
    print(res.groupby(["year", res["result_dt"].dt.month]).size().unstack(0))

    ct = res.groupby(["city", "year"]).size().unstack()
    ct.loc["TOTAL"] = ct.sum()
    ct["growth_pct"] = (ct[2026] / ct[2025] - 1) * 100
    print("\nresults by city:")
    print(ct.round(1))

    pk = res["test_code"].str.startswith("PK").rename("is_package")
    pt = res.groupby([pk, "year"]).size().unstack()
    pt["growth_pct"] = (pt[2026] / pt[2025] - 1) * 100
    print("\npackages vs other tests:")
    print(pt.round(1))

    # bookings side; rescheduled bookings get a new id with no results, so remove them
    resched = book.loc[book["rescheduled_from"].notna(), "booking_id"].unique()
    book_adj = book[~book["booking_id"].isin(resched)]
    bk = book.groupby(["city", "year"]).size().unstack()
    bka = book_adj.groupby(["city", "year"]).size().unstack()
    rr = res.groupby(["city", "year"]).size().unstack()
    chk = pd.DataFrame({
        "booked_growth": (bk[2026] / bk[2025] - 1) * 100,
        "booked_excl_resched_growth": (bka[2026] / bka[2025] - 1) * 100,
        "results_growth": (rr[2026] / rr[2025] - 1) * 100,
        "results_per_booked_2026": rr[2026] / bka[2026],
    })
    print("\nbookings vs results (rescheduled duplicates removed from bookings):")
    print(chk.round(3))


# ------------------------------------------------------------------- main
def clean_all(results_2025, results_2026, bookings_2025, bookings_2026):
    book25 = prep_bookings(bookings_2025, 2025)
    book26 = prep_bookings(bookings_2026, 2026)
    res25 = prep_results(results_2025, book25, 2025)
    res26 = prep_results(results_2026, book26, 2026)
    res = pd.concat([res25, res26], ignore_index=True)
    book = pd.concat([book25, book26], ignore_index=True)
    qa_report(res, book)
    show_growth_trees(book, res)
    return res, book


# ------------------------------------------------------------ growth tree
BRANCHES = ["patients", "visits_per_patient", "tests_per_visit", "results_per_test_booked"]


def growth_tree(book, res, by=None):
    """Results = patients x visits/patient x tests/visit x results/test booked.
    visit = booking. Rescheduled duplicate bookings are removed first (they never get results).
    The four branch growth rates multiply back to the actual growth in results; gap_pp must be 0.

    by = None | 'city' | 'channel' | ['city', 'channel'] -> one tree per segment.
    Results with no matching booking have no channel, so channel splits use matched results
    only; the overall tree (by=None) uses all results."""
    resched = book.loc[book["rescheduled_from"].notna(), "booking_id"].unique()
    b = book[~book["booking_id"].isin(resched)]
    keys = [by] if isinstance(by, str) else list(by or [])
    if not keys:
        b, res = b.assign(_all="ALL"), res.assign(_all="ALL")
        keys = ["_all"]
    else:
        res = res.dropna(subset=keys)

    d = pd.concat({
        "patients": b.groupby(keys + ["year"])["patient_id"].nunique(),
        "visits": b.groupby(keys + ["year"])["booking_id"].nunique(),
        "rows": b.groupby(keys + ["year"]).size(),
        "results": res.groupby(keys + ["year"]).size(),
    }, axis=1).unstack("year")

    def ratio(num, den):
        return (d[num][2026] / d[den][2026]) / (d[num][2025] / d[den][2025])

    g = pd.DataFrame({
        "patients": d["patients"][2026] / d["patients"][2025],
        "visits_per_patient": ratio("visits", "patients"),
        "tests_per_visit": ratio("rows", "visits"),
        "results_per_test_booked": ratio("results", "rows"),
    })
    g["product"] = g.prod(axis=1)
    g["actual"] = d["results"][2026] / d["results"][2025]
    out = ((g - 1) * 100).round(2)
    out["gap_pp"] = (100 * (g["product"] - g["actual"])).round(4)
    out["results_2025"] = d["results"][2025]
    out["results_2026"] = d["results"][2026]
    out.index.names = keys if keys != ["_all"] else ["overall"]
    return out


def contribution(res, by):
    """Additive split: each segment's contribution (percentage points) to total growth.
    Contributions sum to the overall growth. Uses all results; unmatched ones show as NaN."""
    keys = [by] if isinstance(by, str) else list(by)
    t = res.groupby(keys + ["year"], dropna=False).size().unstack("year").fillna(0)
    total25 = t[2025].sum()
    t["growth_pct"] = (t[2026] / t[2025].replace(0, float("nan")) - 1) * 100
    t["contribution_pp"] = (t[2026] - t[2025]) / total25 * 100
    total = pd.DataFrame([[t[2025].sum(), t[2026].sum(), (t[2026].sum() / total25 - 1) * 100,
                           t["contribution_pp"].sum()]], columns=t.columns)
    total.index = (pd.MultiIndex.from_tuples([("TOTAL",) + ("",) * (len(keys) - 1)], names=keys)
                   if len(keys) > 1 else pd.Index(["TOTAL"], name=keys[0]))
    t = pd.concat([t, total])
    return t.round(2)


def show_growth_trees(book, res):
    pd.set_option("display.width", 200)
    pd.set_option("display.max_columns", 20)
    print("\n================ GROWTH TREE: overall ================")
    print(growth_tree(book, res).T)
    for by in ("city", "channel", ["city", "channel"]):
        print(f"\n================ GROWTH TREE by {by} (growth %, gap_pp must be 0) ================")
        print(growth_tree(book, res, by))
        print(f"\n---- contribution to total growth, by {by} ----")
        print(contribution(res, by))
```

```python
import numpy as np, pandas as pd

MONTHS = ["2026-01","2026-02","2026-03","2026-04","2026-05","2026-06"]

def compare_plan(df_plan, res, book):
    """Plan vs actual by city and month, on two bases:
       results = result rows by result month; booked = booked test rows by slot month
       (rescheduled duplicate bookings removed)."""
    p = df_plan.copy()
    p["city"] = p["city"].str.strip().replace({"Bangalore": "Bengaluru"})
    plan = p.set_index("city")[MONTHS]

    r = res[res["year"] == 2026].copy()
    r["m"] = r["result_dt"].dt.strftime("%Y-%m")
    act_res = r[r["m"].isin(MONTHS)].groupby(["city", "m"]).size().unstack(fill_value=0).reindex(columns=MONTHS, fill_value=0)

    resched = book.loc[book["rescheduled_from"].notna(), "booking_id"].unique()
    b = book[(book["year"] == 2026) & ~book["booking_id"].isin(resched)].copy()
    b["m"] = b["slot_dt"].dt.strftime("%Y-%m")
    act_bk = b[b["m"].isin(MONTHS)].groupby(["city", "m"]).size().unstack(fill_value=0).reindex(columns=MONTHS, fill_value=0)

    def table(act, name):
        a = act.reindex(plan.index).fillna(0)
        t = pd.DataFrame({"plan": plan.sum(axis=1), name: a.sum(axis=1)})
        t.loc["TOTAL"] = t.sum()
        t["variance"] = t[name] - t["plan"]
        t["attainment_pct"] = (t[name] / t["plan"] * 100).round(1)
        mon = (a / plan * 100).round(1)
        mon.loc["TOTAL"] = (a.sum() / plan.sum() * 100).round(1)
        return t, mon
    return {"results": table(act_res, "actual_results"), "booked": table(act_bk, "actual_booked")}


def show_plan_comparison(df_plan, res, book):
    pd.set_option("display.width", 200)
    out = compare_plan(df_plan, res, book)
    for name, (tbl, monthly) in out.items():
        print(f"\n===== plan vs actual: {name} =====")
        print(tbl)
        print(f"\nattainment % by month ({name}):")
        print(monthly)
    return out
```
