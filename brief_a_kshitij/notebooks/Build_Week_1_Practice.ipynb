```python
import pandas as pd

def recover_missing_test_codes(bookings_df, results_df):
    """
    Recover missing test_code values when exactly one booked code
    is absent from the recorded results for that booking.

    Returns:
        cleaned_results: copy of results with inferred codes filled
        audit: DataFrame showing each missing row and its status
    """
    cleaned_results = results_df.copy()

    # Identify rows with missing test codes
    missing_mask = cleaned_results["test_code"].isna()

    if not missing_mask.any():
        return cleaned_results, pd.DataFrame()

    missing_ids = cleaned_results.loc[
        missing_mask, "booking_id"
    ].unique()

    # Unique booked codes for affected bookings
    booked_codes = (
        bookings_df[
            bookings_df["booking_id"].isin(missing_ids)
        ]
        .dropna(subset=["test_code"])
        .groupby("booking_id")["test_code"]
        .agg(lambda s: set(s))
    )

    # Existing recorded result codes for affected bookings
    recorded_codes = (
        cleaned_results[
            cleaned_results["booking_id"].isin(missing_ids)
        ]
        .dropna(subset=["test_code"])
        .groupby("booking_id")["test_code"]
        .agg(lambda s: set(s))
    )

    audit_rows = []

    for idx in cleaned_results.index[missing_mask]:
        booking_id = cleaned_results.at[idx, "booking_id"]

        booked = booked_codes.get(booking_id, set())
        recorded = recorded_codes.get(booking_id, set())

        candidates = booked - recorded

        if len(candidates) == 1:
            inferred_code = next(iter(candidates))
            cleaned_results.at[idx, "test_code"] = inferred_code
            status = "inferred_unique_candidate"
        else:
            inferred_code = None
            status = (
                "ambiguous"
                if len(candidates) > 1
                else "no_candidate"
            )

        audit_rows.append({
            "row_index": idx,
            "booking_id": booking_id,
            "inferred_test_code": inferred_code,
            "candidate_codes": sorted(candidates),
            "status": status
        })

    audit = pd.DataFrame(audit_rows)

    return cleaned_results, audit
```

```python
import pandas as pd

bookings = df_bookings_ahmedabad_2026_06.copy()
results = df_results_ahmedabad_2026_06

# Parse dates
bookings['slot_at'] = pd.to_datetime(
    bookings['slot_at'], errors='coerce', format='mixed', dayfirst=True
)
results['result_ts'] = pd.to_datetime(
    results['result_ts'], errors='coerce', format='mixed'
)

# Keep only Result rows with missing test_code
missing = results[results['test_code'].isna()].copy()
missing['_result_idx'] = missing.index

# Match missing Result rows to Booking by booking_id
candidates = missing.merge(
    bookings[['booking_id', 'test_code', 'slot_at', 'source']],
    on='booking_id',
    how='left',
    suffixes=('_result', '_booking')
)

# Apply date rules
result_date = candidates['result_ts'].dt.normalize()
slot_date = candidates['slot_at'].dt.normalize()
source = candidates['source'].str.strip().str.lower()

valid_date = (
    (source.eq('walkin') & (slot_date == result_date)) |
    (source.isin(['app', 'phone']) &
     (slot_date + pd.Timedelta(days=1) == result_date))
)

matched = candidates.loc[valid_date].dropna(
    subset=['test_code_booking']
).copy()

# Existing non-null Result test codes for each booking
existing_codes = (
    results.dropna(subset=['test_code'])
    .groupby('booking_id')['test_code']
    .agg(set)
    .to_dict()
)

# Candidate codes not already present in Result
matched['existing_codes'] = matched['booking_id'].map(
    existing_codes
).apply(lambda x: x if isinstance(x, set) else set())

matched['unused_codes'] = matched.apply(
    lambda row: sorted(
        set([row['test_code_booking']]) - row['existing_codes']
    ),
    axis=1
)

# Keep candidates not already used
matched = matched[
    matched['unused_codes'].map(len) == 1
].copy()

matched['proposed_code'] = matched['unused_codes'].str[0]

# Only fill when there is exactly one distinct proposed code
# for that missing Result row
proposals = (
    matched.groupby('_result_idx')['proposed_code']
    .agg(lambda x: set(x))
)

safe_proposals = proposals[proposals.map(len) == 1].map(
    lambda x: next(iter(x))
)

# Update only the originally missing test_code values
before = results['test_code'].isna().sum()

for idx, code in safe_proposals.items():
    if pd.isna(results.at[idx, 'test_code']):
        results.at[idx, 'test_code'] = code

after = results['test_code'].isna().sum()

print("Filled test codes:", before - after)
print("Remaining null test codes:", after)
```
