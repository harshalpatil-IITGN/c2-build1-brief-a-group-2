```python
# =====================================================================
# KALPA HEALTH: invoices -> collections -> settlements bridge
# Needs these DataFrames loaded: df_invoices_YYYYQn, df_gateway_export_YYYYQn,
# df_settlements_YYYYQn. Run top to bottom in one cell, after a fresh restart.
# =====================================================================
import re, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
warnings.filterwarnings("ignore")

# ---------- 0. SETTINGS ----------
QUARTER = "2026Q2"   # quarter for the brief's bridge (section 10)
BASIS   = "ist"      # payment quarter from the IST timestamp ("ist") or from the gateway file ("file")
TOL     = 0.5        # rupees allowed when two amounts are called equal

INV_COLS = ["invoice_id", "invoice_date", "booking_id", "patient_id", "city", "channel", "amount"]
GW_COLS  = ["payment_id", "notes_invoice_ref", "amount_paise", "status", "created_at_epoch", "method", "settlement_id"]
ST_COLS  = ["settlement_id", "amount_paise", "fees_paise", "tax_paise", "utr", "created_at_epoch", "status"]

# ---------- 1. HELPERS ----------
def clean_invoice_id(value):
    """Uppercase, strip non-alphanumerics, rebuild INV-YYYY-NNNNNN. None if it cannot be read."""
    if pd.isna(value):
        return None
    s = re.sub(r"[^A-Za-z0-9]", "", str(value)).upper()
    m = re.fullmatch(r"INV(\d{4})(\d{6})", s)
    return f"INV-{m.group(1)}-{m.group(2)}" if m else None

def offset_fix(value):
    """Payment refs that carry the invoice number + 900,000 (6 or 7 digits). Returns the corrected id or None."""
    if pd.isna(value):
        return None
    s = re.sub(r"[^A-Za-z0-9]", "", str(value)).upper()
    m = re.fullmatch(r"INV(\d{4})(\d{6,7})", s)
    if not m or int(m.group(2)) < 900000:
        return None
    return f"INV-{m.group(1)}-{int(m.group(2)) - 900000:06d}"

def find_frames(prefix):
    out = {}
    for name, obj in list(globals().items()):
        m = re.fullmatch(rf"{prefix}_(\d{{4}}Q[1-4])", name)
        if m and isinstance(obj, pd.DataFrame):
            out[m.group(1)] = obj
    return dict(sorted(out.items()))

def tidy(df):
    df = df.copy()
    for c in df.columns:
        if isinstance(df[c].dtype, pd.PeriodDtype):
            df[c] = df[c].astype(str)
    return df

# ---------- 2. LOAD, COMBINE, CLEAN ----------
inv_frames, gw_frames, st_frames = find_frames("df_invoices"), find_frames("df_gateway_export"), find_frames("df_settlements")
print("invoices:", list(inv_frames)); print("gateway:", list(gw_frames)); print("settlements:", list(st_frames))

I = pd.concat([d[INV_COLS].assign(inv_file=q) for q, d in inv_frames.items()], ignore_index=True)
G = pd.concat([d[GW_COLS].assign(gw_file=q) for q, d in gw_frames.items()], ignore_index=True)
S = pd.concat([d[ST_COLS].assign(st_file=q) for q, d in st_frames.items()], ignore_index=True)

I["ref_clean"] = I["invoice_id"].apply(clean_invoice_id)
I["amount_rs"] = pd.to_numeric(I["amount"].astype(str).str.replace(r"[^\d.\-]", "", regex=True), errors="coerce")
I["inv_date"]  = pd.to_datetime(I["invoice_date"], errors="coerce")

to_ist = lambda e: pd.to_datetime(e, unit="s", utc=True).dt.tz_convert("Asia/Kolkata").dt.tz_localize(None)
G["paid_at"] = to_ist(G["created_at_epoch"])
S["created"] = to_ist(S["created_at_epoch"])

assert I["ref_clean"].notna().all(), "unreadable invoice ids"
assert not I["ref_clean"].duplicated().any(), "duplicate invoice ids after cleaning"
assert not G["payment_id"].duplicated().any() and not S["settlement_id"].duplicated().any()
print("bad amounts:", int(I["amount_rs"].isna().sum()), "| bad dates:", int(I["inv_date"].isna().sum()))

# ---------- 3. PROFILE ----------
print(I.groupby("inv_file").agg(rows=("ref_clean", "size"), rs=("amount_rs", "sum"), first=("inv_date", "min"), last=("inv_date", "max")))
print(G.groupby(["gw_file", "status"]).size().unstack(fill_value=0))
print(pd.crosstab(G["status"], G["settlement_id"].notna()))          # only captured payments are settled
print("epoch digits:", G["created_at_epoch"].astype(str).str.len().value_counts().to_dict())

# ---------- 4. PAYMENT REFERENCES: clean, then the +900,000 fix (validated) ----------
G["ref_clean"] = G["notes_invoice_ref"].apply(clean_invoice_id)
G["ref_fixed"] = False
refs = set(I["ref_clean"])
miss = ~G["ref_clean"].isin(refs)
print("payments matching no invoice before the fix:", int(miss.sum()), "| captured:", int((miss & (G["status"] == "captured")).sum()))

inv_idx = I.set_index("ref_clean")[["amount_rs", "inv_date"]]
need = G[miss].copy()
need["ref_try"] = need["notes_invoice_ref"].apply(offset_fix)
need = need.dropna(subset=["ref_try"]).join(inv_idx, on="ref_try")
need["ok"] = (((need["amount_paise"] / 100 - need["amount_rs"]).abs() < TOL)
              & ((need["paid_at"] - need["inv_date"]).dt.days.between(0, 15)))      # amount AND date must agree
need["shape"] = np.where(need["notes_invoice_ref"].astype(str).str.replace(r"[^A-Za-z0-9]", "", regex=True).str.len() == 14,
                         "7-digit", "6-digit 9xxxxx")
print(need.groupby(["shape", "status"]).agg(payments=("ok", "size"), validated=("ok", "sum")))
fixed = need[need["ok"]]
G.loc[fixed.index, "ref_clean"] = fixed["ref_try"]
G.loc[fixed.index, "ref_fixed"] = True                                # only validated payments are changed
print("fixed:", len(fixed), "| captured Rs", fixed.loc[fixed["status"] == "captured", "amount_paise"].sum() / 100)

# ---------- 5. CLASSIFY EVERY INVOICE ----------
cap = G[G["status"] == "captured"].copy()
cap["amt_rs"] = cap["amount_paise"] / 100
cap["pq"] = cap["gw_file"] if BASIS == "file" else cap["paid_at"].dt.to_period("Q").astype(str)

pay = cap.groupby("ref_clean").agg(n_cap=("payment_id", "size"), paid_rs=("amt_rs", "sum"), first_paid=("paid_at", "min"))
M = I.copy().join(pay, on="ref_clean")
M["n_cap"] = M["n_cap"].fillna(0).astype(int)
M["paid_rs"] = M["paid_rs"].fillna(0.0)
M["diff_rs"] = (M["paid_rs"] - M["amount_rs"]).round(2)
M["bucket"] = np.select([M["n_cap"] == 0, M["diff_rs"].abs() < TOL, M["diff_rs"] > 0],
                        ["no capture", "paid in full", "overpaid"], default="underpaid")
M["inv_q"], M["paid_q"] = M["inv_date"].dt.to_period("Q"), M["first_paid"].dt.to_period("Q")
assert len(M) == len(I)

orph = cap[~cap["ref_clean"].isin(refs)]                              # captured, reference matches no invoice
print(M.groupby("bucket").agg(invoices=("ref_clean", "size"), invoice_rs=("amount_rs", "sum"),
                              paid_rs=("paid_rs", "sum"), net_diff=("diff_rs", "sum")))
print("payments with no invoice after the fix:", len(orph), "| Rs", orph["amt_rs"].sum())

# ---------- 6. EXTRA CAPTURES (second or later payment on an invoice) ----------
multi = (cap[cap["ref_clean"].isin(M.loc[M["n_cap"] >= 2, "ref_clean"])]
           .merge(I[["ref_clean", "patient_id", "amount_rs"]], on="ref_clean"))
multi["eq_inv"] = (multi["amt_rs"] - multi["amount_rs"]).abs() < TOL
extra_diff = multi[~multi["eq_inv"]]                                  # a payment of some other amount
ed = (multi[multi["eq_inv"]].groupby("ref_clean").agg(n=("payment_id", "size"), amt=("amt_rs", "first"))
        .query("n >= 2"))                                             # the invoice amount paid twice
exact_extra_rs = ((ed["n"] - 1) * ed["amt"]).sum()
print("extra captures, different amount:", len(extra_diff), "| Rs", extra_diff["amt_rs"].sum())
print("exact duplicate invoices:", len(ed), "| Rs extra", exact_extra_rs)
print(M[M["bucket"].isin(["overpaid", "underpaid"])].groupby(["bucket", "n_cap"])
        .agg(invoices=("ref_clean", "size"), diff=("diff_rs", "sum")))

# ---------- 7. TIMING AND THE UNPAID WINDOW ----------
paid = M[M["n_cap"] > 0].copy()
paid["lag_days"] = (paid["first_paid"] - paid["inv_date"]).dt.days
print(paid["lag_days"].quantile([.5, .9, .99, 1]).to_dict(), "| max lag by invoice quarter:",
      paid.groupby("inv_q")["lag_days"].max().to_dict())
window = int(paid["lag_days"].max())                                  # observed maximum, not a chosen percentile
data_end = G["paid_at"].max()

nc = M[M["bucket"] == "no capture"].copy()
nc["age_days"] = (data_end - nc["inv_date"]).dt.days
nc["class"] = np.where(nc["age_days"] <= window, "recent: may still be paid", "aged: treat as unpaid")
M["nc_class"] = M["ref_clean"].map(nc.set_index("ref_clean")["class"])
print("window (days):", window, "| data ends:", data_end)
print(nc.groupby("class").agg(invoices=("ref_clean", "size"), rs=("amount_rs", "sum")))
print(nc.groupby(["inv_q", "class"]).size().unstack(fill_value=0))
late_q = (M["n_cap"] > 0) & (M["paid_q"] > M["inv_q"])
print("paid in a later quarter:", int(late_q.sum()), "| Rs", M.loc[late_q, "amount_rs"].sum())

# ---------- 8. SETTLEMENTS ----------
capg = cap.groupby("settlement_id")["amount_paise"].agg(cap_paise="sum", n_pay="size")
R = S.set_index("settlement_id").join(capg, how="outer")
Rs_ = R[R["amount_paise"].notna()]
assert ((Rs_["amount_paise"] + Rs_["fees_paise"] + Rs_["tax_paise"]) == Rs_["cap_paise"]).all(), "amount + fees + tax != captured"
print("settlements tie (amount + fees + tax = captured):", len(Rs_), "of", len(S), "| batches without a settlement row:", int(R["amount_paise"].isna().sum()))
print("fee share:", (Rs_["fees_paise"] / Rs_["cap_paise"]).round(4).unique()[:3], "| tax share of fee:", (Rs_["tax_paise"] / Rs_["fees_paise"]).round(3).unique()[:3])
gap_h = (S.set_index("settlement_id")["created"] - cap.groupby("settlement_id")["paid_at"].max()).dt.total_seconds().div(3600).dropna()
print("hours from a batch's last payment to its creation:\n", gap_h.describe().round(2).to_dict())

# allocate each settlement back to its payments, then to invoices (pro rata; exact because fee is a fixed %)
st = S.set_index("settlement_id")
cap["batch_rs"] = cap.groupby("settlement_id")["amt_rs"].transform("sum")
cap["net_rs"] = cap["amt_rs"] / cap["batch_rs"] * cap["settlement_id"].map(st["amount_paise"]) / 100   # NaN = open batch
cap["settled_on"] = cap["settlement_id"].map(st["created"])
cap["unsettled_rs"] = np.where(cap["net_rs"].isna(), cap["amt_rs"], 0.0)
assert abs(cap["net_rs"].sum() - S["amount_paise"].sum() / 100) < 1, "allocation does not add back to settlements"

per_inv = cap.groupby("ref_clean").agg(net_rs=("net_rs", "sum"), unsettled_rs=("unsettled_rs", "sum"),
                                       settled_on=("settled_on", "max"))
M = M.join(per_inv, on="ref_clean")
M["net_rs"], M["unsettled_rs"] = M["net_rs"].fillna(0.0), M["unsettled_rs"].fillna(0.0)
M["fees_tax"] = M["paid_rs"] - M["unsettled_rs"] - M["net_rs"]
M["settled_q"] = M["settled_on"].dt.to_period("Q")
print("invoices with money not yet settled:", int((M["unsettled_rs"] > 0).sum()), "| Rs", M.loc[M["unsettled_rs"] > 0, "amount_rs"].sum())
print("credited in a later quarter than invoiced:", int(((M["n_cap"] > 0) & (M["settled_q"] > M["inv_q"])).sum()))

# net cash vs invoiced, by invoice quarter, split into its causes
cls = M["nc_class"].fillna("")
M["unpaid_aged"]   = np.where(cls.str.startswith("aged"), -M["diff_rs"], 0.0)
M["unpaid_recent"] = np.where(cls.str.startswith("recent"), -M["diff_rs"], 0.0)
M["under_over"]    = np.where(M["bucket"].isin(["underpaid", "overpaid"]), -M["diff_rs"], 0.0)
cols = ["unpaid_aged", "unpaid_recent", "under_over", "unsettled_rs", "fees_tax"]
gq = M.groupby("inv_q")[["amount_rs", "net_rs"] + cols].sum()
gq["gap"] = gq["amount_rs"] - gq["net_rs"]
gq["net_share"] = (gq["net_rs"] / gq["amount_rs"]).round(4)
assert (gq["gap"] - gq[cols].sum(axis=1)).abs().max() < 1
print((gq[["amount_rs", "net_rs", "gap"] + cols] / 1e5).round(1), "\n(Rs lakh)  net share:", gq["net_share"].to_dict())

# ---------- 9. THE BRIDGE (assertions: it must tie at every step) ----------
inv_total = M["amount_rs"].sum()
aged_m, rec_m = nc["class"].str.startswith("aged"), nc["class"].str.startswith("recent")
aged_rs, recent_rs = nc.loc[aged_m, "amount_rs"].sum(), nc.loc[rec_m, "amount_rs"].sum()
under = -M.loc[M["bucket"] == "underpaid", "diff_rs"].sum()
over  = M.loc[M["bucket"] == "overpaid", "diff_rs"].sum()
matched, orph_rs, all_cap = M["paid_rs"].sum(), orph["amt_rs"].sum(), cap["amt_rs"].sum()
fees, tax = S["fees_paise"].sum() / 100, S["tax_paise"].sum() / 100
unsettled, settled = cap["unsettled_rs"].sum(), S["amount_paise"].sum() / 100

assert abs(inv_total - aged_rs - recent_rs - under + over - matched) < 1, "invoice side does not tie"
assert abs(matched + orph_rs - all_cap) < 1, "a captured payment was dropped in the join"
assert abs(all_cap - fees - tax - unsettled - settled) < 1, "settlement side does not tie"
assert len(M) == int(aged_m.sum() + rec_m.sum() + (M["n_cap"] > 0).sum())

bridge = pd.DataFrame([
    ("Invoiced",                           len(M),                          "invoices",    inv_total),
    ("Never paid (aged)",                  int(aged_m.sum()),               "invoices",   -aged_rs),
    ("Recent, not yet paid",               int(rec_m.sum()),                "invoices",   -recent_rs),
    ("Underpaid shortfall",                int((M["bucket"] == "underpaid").sum()), "invoices", -under),
    ("Overpaid excess",                    int((M["bucket"] == "overpaid").sum()),  "invoices",  over),
    ("Payments with no invoice",           len(orph),                       "payments",    orph_rs),
    ("Captured",                           len(cap),                        "payments",    all_cap),
    ("Gateway fees",                       len(Rs_),                        "settlements", -fees),
    ("Tax on fees",                        len(Rs_),                        "settlements", -tax),
    ("Open batch, not yet settled",        int((cap["unsettled_rs"] > 0).sum()), "payments", -unsettled),
    ("Settled",                            len(Rs_),                        "settlements", settled)],
    columns=["bar", "count", "unit", "rs"])
bridge["rs_crore"] = (bridge["rs"] / 1e7).round(4)
print(bridge.to_string(index=False)); print("bridge balances")

# ---------- 10. THE BRIEF'S BRIDGE FOR ONE QUARTER ----------
def quarter_bridge(Q):
    feed = cap[cap["pq"] == Q]
    iq = M.loc[M["inv_file"] == Q, ["ref_clean", "amount_rs"]].set_index("ref_clean")
    when = np.where(cap["pq"] < Q, "p_early", np.where(cap["pq"] == Q, "p_in", "p_late"))
    pp = (cap.assign(when=when)
             .pivot_table(index="ref_clean", columns="when", values="amt_rs", aggfunc="sum", fill_value=0)
             .reindex(columns=["p_early", "p_in", "p_late"], fill_value=0))
    iq = iq.join(pp).fillna(0)
    none = iq[["p_early", "p_in", "p_late"]].sum(axis=1) == 0           # no captured payment anywhere
    inq = iq["p_in"] > 0                                                 # paid in the quarter
    out = ~none & ~inq                                                   # paid, but not in this quarter
    X, no_pay, out_rs = iq["amount_rs"].sum(), iq.loc[none, "amount_rs"].sum(), iq.loc[out, "amount_rs"].sum()
    diff = (iq.loc[inq, "p_in"] - iq.loc[inq, "amount_rs"]).sum()
    earlier = feed[~feed["ref_clean"].isin(iq.index)]                    # feed money for other quarters' invoices
    Y, t_in = feed["amt_rs"].sum(), earlier["amt_rs"].sum()
    assert abs(X - no_pay - out_rs + diff + t_in - Y) < 1, f"{Q}: bridge does not tie"
    n_diff = int((inq & ((iq["p_in"] - iq["amount_rs"]).abs() >= TOL)).sum())
    b = pd.DataFrame({"bar": ["Invoiced in the quarter", "Invoices with no payment", "Paid after the quarter closed",
                              "Collected at a different amount", "Earlier invoices paid this quarter", "Collected in the payment feed"],
                      "count": [len(iq), int(none.sum()), int(out.sum()), n_diff, len(earlier), len(feed)],
                      "unit": ["invoices"] * 4 + ["payments"] * 2,
                      "rs": [X, -no_pay, -out_rs, diff, t_in, Y]})
    split = M[M["ref_clean"].isin(iq.index[none])].groupby("nc_class").agg(invoices=("ref_clean", "size"), rs=("amount_rs", "sum"))
    return b, split, int((out & (iq["p_late"] == 0)).sum())

qb, qsplit, n_before_only = quarter_bridge(QUARTER)
print(f"\nBRIDGE FOR {QUARTER} (payment quarter basis: {BASIS})"); print(qb.round(0).to_string(index=False))
print(qsplit); print("invoices paid only BEFORE the quarter:", n_before_only)
allq = {Q: quarter_bridge(Q)[0].set_index("bar")["rs"] for Q in sorted(M["inv_file"].unique())}
print((pd.DataFrame(allq) / 1e5).round(1))                              # Rs lakh, every quarter

# ---------- 11. WEEKLY CHECK ----------
THRESH = {"ratio_min": 0.90, "unpaid_max": 0.036, "fix_band": (0.005, 0.02), "extra_max": 0.007}
wk = lambda s: s.dt.to_period("W-SUN")

def weekly_table():
    d = M[["inv_date", "first_paid"]].copy()
    d["w"] = wk(d["inv_date"])
    d["unpaid_w"] = d["first_paid"].isna() | ((d["first_paid"] - d["inv_date"]).dt.days > window)
    W = d.groupby("w").agg(invoices=("inv_date", "size"), unpaid_at_window=("unpaid_w", "mean"))
    c = cap.sort_values("paid_at").copy()
    c["w"] = wk(c["paid_at"])
    c["unmatched"] = ~c["ref_clean"].isin(refs)
    c["extra"] = c["ref_clean"].duplicated() & c["ref_clean"].notna()
    P = c.groupby("w").agg(captures=("payment_id", "size"), fix_rate=("ref_fixed", "mean"),
                           unmatched=("unmatched", "sum"), extra_rate=("extra", "mean"))
    s = st.join(cap.groupby("settlement_id")["amount_paise"].sum().rename("cap_paise"))
    s["bad"] = ((s["amount_paise"] + s["fees_paise"] + s["tax_paise"] != s["cap_paise"])
                | ((s["fees_paise"] / s["cap_paise"] - 0.02).abs() > 1e-4))
    s["w"] = wk(s["created"])
    W = W.join(P).join(s.groupby("w")["bad"].sum().rename("bad_settlements"))
    W["bad_settlements"] = W["bad_settlements"].fillna(0)
    W["capture_ratio"] = W["captures"] / W["invoices"]
    return W[W.index >= pd.Period("2025-01-06", "W-SUN")]               # drop the partial first week

W = weekly_table()
as_of = (data_end - pd.Timedelta(hours=24)).normalize()
done = [w for w in W.index if w.end_time.normalize() <= as_of]
last = done[-1]
coh = [w for w in done if w.end_time.normalize() + pd.Timedelta(days=window) <= as_of][-1]
med = W.loc[done[-13:-1], "capture_ratio"].median()
r, c0 = W.loc[last], W.loc[coh]
stuck = int((~cap["settlement_id"].isin(S["settlement_id"]) & (cap["paid_at"] < data_end - pd.Timedelta(hours=36))).sum())
ties = abs(matched + orph_rs - all_cap) < 1 and abs(all_cap - fees - tax - unsettled - settled) < 1
rows = [
 (1, f"captures / invoices raised, week {last}", f"{r['capture_ratio']:.3f}", f"median {med:.3f}", r["capture_ratio"] < THRESH["ratio_min"] * med),
 (2, f"unpaid after {window} days, cohort {coh}", f"{c0['unpaid_at_window']:.2%}", "about 2.9%", c0["unpaid_at_window"] > THRESH["unpaid_max"]),
 (3, f"refs needing the +900,000 fix, week {last}", f"{r['fix_rate']:.2%}", "about 1.0%", not (THRESH["fix_band"][0] <= r["fix_rate"] <= THRESH["fix_band"][1])),
 (4, f"payments with no invoice, week {last}", int(r["unmatched"]), "0", r["unmatched"] > 0),
 (5, f"extra-capture rate, week {last}", f"{r['extra_rate']:.2%}", "about 0.4%", r["extra_rate"] > THRESH["extra_max"]),
 (6, f"settlements failing amount + fees + tax = captured, week {last}", int(r["bad_settlements"]), "0", r["bad_settlements"] > 0),
 (7, "captured payments older than 36h with no settlement row", stuck, "0", stuck > 0),
 (8, "rupee ties hold", str(ties), "True", not ties)]
weekly = pd.DataFrame(rows, columns=["#", "check", "value", "normal", "alert"])
weekly["status"] = np.where(weekly["alert"], "ALERT", "ok")
print("\nWEEKLY CHECK, as of", as_of.date()); print(weekly.drop(columns="alert").to_string(index=False))
Wd = W.loc[done].copy()
Wd["ratio_alert"] = Wd["capture_ratio"] < THRESH["ratio_min"] * Wd["capture_ratio"].rolling(12).median().shift(1)
Wd["window_alert"] = ((Wd["unpaid_at_window"] > THRESH["unpaid_max"])
                      & np.array([w.end_time.normalize() + pd.Timedelta(days=window) <= as_of for w in Wd.index]))
print("weeks the rules would have flagged (backtest):\n",
      Wd.loc[Wd[["ratio_alert", "window_alert"]].any(axis=1), ["invoices", "captures", "capture_ratio", "unpaid_at_window"]])

# ---------- 12. CHARTS ----------
def waterfall(ax, labels, values, kinds, lab, title):
    run, tops = 0, []
    for i, (v, k) in enumerate(zip(values, kinds)):
        if k == "total":
            ax.bar(i, v, color="#3b4a8c"); run, y = v, v
        else:
            ax.bar(i, v, bottom=run, color="#2e8b57" if v > 0 else "#c0392b"); y = max(run, run + v); run += v
        tops.append(run); ax.text(i, y, lab(v, k), ha="center", va="bottom", fontsize=8)
    lo, hi = min(tops), max(tops); pad = (hi - lo) * 0.2 or 1
    ax.set_ylim(lo - pad, hi + pad); ax.set_title(title, fontsize=10)
    ax.set_xticks(range(len(labels))); ax.set_xticklabels(labels, rotation=35, ha="right", fontsize=8)

def rs_label(v, k):
    a = abs(v); s = "" if k == "total" else ("+" if v > 0 else "-")
    return s + (f"Rs {a:,.2f} cr" if a >= 1 else f"Rs {a * 100:,.1f} L")

steps = [("Invoiced", inv_total, "total"), ("Never paid\n(aged)", -aged_rs, "delta"), ("Recent,\nnot yet paid", -recent_rs, "delta"),
         ("Underpaid", -under, "delta"), ("Overpaid", over, "delta")]
if abs(orph_rs) >= 1:
    steps.append(("Payments,\nno invoice", orph_rs, "delta"))
steps += [("Captured", all_cap, "total"), ("Gateway\nfees (2%)", -fees, "delta"), ("Tax on\nfees", -tax, "delta"),
          ("Open batch", -unsettled, "delta"), ("Settled", settled, "total")]
n_inv, n_cap_inv = len(M), int((M["n_cap"] > 0).sum())

fig, (a1, a2) = plt.subplots(1, 2, figsize=(16, 5.5), gridspec_kw={"width_ratios": [1, 2.6]})
waterfall(a1, ["Invoices", "Never paid\n(aged)", "Recent,\nnot yet paid", "Invoices with\na capture"],
          [n_inv, -int(aged_m.sum()), -int(rec_m.sum()), n_cap_inv], ["total", "delta", "delta", "total"],
          lambda v, k: ("" if k == "total" else ("+" if v > 0 else "-")) + f"{abs(v):,.0f}", "Counts (axis truncated)")
waterfall(a2, [s[0] for s in steps], [s[1] / 1e7 for s in steps], [s[2] for s in steps], rs_label,
          "Rupees: invoiced to settled (axis truncated)")
plt.tight_layout(); fig.savefig("bridge_full.png", dpi=200, bbox_inches="tight"); plt.show()

fig, ax = plt.subplots(figsize=(11, 5.5))
waterfall(ax, [f"{b}\n{n:,} {u}" for b, n, u in zip(qb["bar"], qb["count"], qb["unit"])],
          (qb["rs"] / 1e7).tolist(), ["total", "delta", "delta", "delta", "delta", "total"], rs_label,
          f"{QUARTER}: invoiced to collected (axis truncated)")
plt.tight_layout(); fig.savefig(f"bridge_{QUARTER}.png", dpi=200, bbox_inches="tight"); plt.show()

# ---------- 13. LISTS FINANCE CAN ACT ON ----------
base = ["ref_clean", "patient_id", "city", "channel", "inv_date", "amount_rs", "age_days"]
lists = {
 "unpaid_aged":      nc[nc["class"].str.startswith("aged")][base].sort_values("inv_date"),
 "recent_unpaid":    nc[nc["class"].str.startswith("recent")][base].sort_values("inv_date"),
 "extra_captures":   extra_diff[["ref_clean", "payment_id", "paid_at", "method", "amt_rs", "amount_rs"]]
                       .rename(columns={"amt_rs": "captured_rs", "amount_rs": "invoice_rs"}),
 "exact_duplicates": ed.reset_index(),
 "amount_mismatch":  M[M["bucket"].isin(["underpaid", "overpaid"]) & (M["n_cap"] == 1)][["ref_clean", "amount_rs", "paid_rs", "diff_rs"]],
 "recovered_refs":   fixed[["payment_id", "notes_invoice_ref", "ref_try", "status", "amount_paise", "paid_at"]],
 "open_batch":       cap[cap["unsettled_rs"] > 0][["payment_id", "ref_clean", "amt_rs", "paid_at", "settlement_id"]]}
try:
    with pd.ExcelWriter("finance_action_lists.xlsx") as xw:
        for k, v in lists.items():
            tidy(v).to_excel(xw, sheet_name=k[:31], index=False)
    print("saved finance_action_lists.xlsx:", {k: len(v) for k, v in lists.items()})
except Exception as e:
    for k, v in lists.items():
        tidy(v).to_csv(f"finance_{k}.csv", index=False)
    print("Excel writer unavailable, saved CSVs instead:", e)

# ---------- 14. SELF-CHECK against the numbers we reached in this chat ----------
def expect(name, got, want):
    print(f"{'OK  ' if abs(got - want) < 1 else 'DIFF'} {name}: {got:,.2f} (expected {want:,.2f})")
print("\nSELF-CHECK")
expect("invoices", len(M), 470776)
expect("paid in full", (M["bucket"] == "paid in full").sum(), 449799)
expect("never paid (aged) invoices", aged_m.sum(), 13357)
expect("recent unpaid invoices", rec_m.sum(), 5607)
expect("underpaid invoices", (M["bucket"] == "underpaid").sum(), 38)
expect("overpaid invoices", (M["bucket"] == "overpaid").sum(), 1975)
expect("extra captures, different amount", len(extra_diff), 1923)
expect("exact duplicate invoices", len(ed), 18)
expect("payments fixed by +900,000 (captured)", (fixed["status"] == "captured").sum(), 4566)
expect("captured payments", len(cap), 453753)
expect("payments with no invoice", len(orph), 0)
expect("invoiced Rs", inv_total, 737020069)
expect("captured Rs", all_cap, 709568514)
expect("settled Rs", settled, 692164108.35)
expect("fees Rs", fees, 14177880.14)
expect("paid in a later quarter", late_q.sum(), 15293)
expect("settlements", len(S), 547)
```
