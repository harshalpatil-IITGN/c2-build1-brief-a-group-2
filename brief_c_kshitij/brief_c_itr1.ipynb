```python
import re
import pandas as pd

def clean_invoice_id(value):
    """Standardise to INV-YYYY-NNNNNN. Returns None if it can't be parsed."""
    if pd.isna(value):
        return None
    # Uppercase and strip everything except letters and digits
    s = re.sub(r"[^A-Za-z0-9]", "", str(value)).upper()
    m = re.fullmatch(r"INV(\d{4})(\d{6})", s)
    return f"INV-{m.group(1)}-{m.group(2)}" if m else None
```

```python
import re
import numpy as np
import pandas as pd

def clean_invoice_id(value):
    if pd.isna(value):
        return None
    s = re.sub(r"[^A-Za-z0-9]", "", str(value)).upper()
    m = re.fullmatch(r"INV(\d{4})(\d{6})", s)
    return f"INV-{m.group(1)}-{m.group(2)}" if m else None

def find_frames(prefix):
    out = {}
    for name, obj in list(globals().items()):
        m = re.fullmatch(rf"{prefix}_(\d{{4}}Q[1-4])", name)
        if m and isinstance(obj, pd.DataFrame):
            out[m.group(1)] = obj
    return dict(sorted(out.items()))
```

```python
THRESH = {"ratio_min": 0.90, "unpaid15_max": 0.036, "fix_band": (0.005, 0.02), "extra_max": 0.007}

def weekly_table(M, G, S):
    refs = set(M["ref_clean"])
    cap = G[G["status"] == "captured"].copy()
    wk = lambda s: s.dt.to_period("W-SUN")                      # weeks run Monday to Sunday

    inv = M[["inv_date", "first_paid"]].copy()
    inv["w"] = wk(inv["inv_date"])
    inv["unpaid15"] = inv["first_paid"].isna() | ((inv["first_paid"] - inv["inv_date"]).dt.days > 15)
    W = inv.groupby("w").agg(invoices=("inv_date", "size"), unpaid15=("unpaid15", "mean"))

    cap["w"] = wk(cap["paid_at"])
    cap["needs_fix"] = ~cap["notes_invoice_ref"].apply(clean_invoice_id).isin(refs)
    cap["unmatched"] = ~cap["ref_clean"].isin(refs)
    cap = cap.sort_values("paid_at")
    cap["extra"] = cap.groupby("ref_clean", dropna=False).cumcount() >= 1
    P = cap.groupby("w").agg(captures=("payment_id", "size"), fix_rate=("needs_fix", "mean"),
                             unmatched=("unmatched", "sum"), extra_rate=("extra", "mean"))
    W = W.join(P)
    W["capture_ratio"] = W["captures"] / W["invoices"]

    capg = cap.groupby("settlement_id")["amount_paise"].sum().rename("cap_paise")
    s = S.set_index("settlement_id").join(capg)
    s["bad"] = ((s["amount_paise"] + s["fees_paise"] + s["tax_paise"] != s["cap_paise"])
                | ((s["fees_paise"] / s["cap_paise"] - 0.02).abs() > 1e-4))
    s["w"] = wk(s["created"])
    W = W.join(s.groupby("w")["bad"].sum().rename("bad_settlements"))
    W = W[W["invoices"] > 1000]                                # drop the sparse 2024 weeks

    data_end = G["paid_at"].max()
    stuck = int((~cap["settlement_id"].isin(S["settlement_id"])
                 & (cap["paid_at"] < data_end - pd.Timedelta(hours=36))).sum())
    unsettled_rs = cap.loc[~cap["settlement_id"].isin(S["settlement_id"]), "amount_paise"].sum() / 100
    return W, cap, stuck, unsettled_rs, data_end

def weekly_check(M, G, S):
    W, cap, stuck, unsettled_rs, data_end = weekly_table(M, G, S)
    as_of = (data_end - pd.Timedelta(hours=24)).normalize()         # last complete day
    done = [w for w in W.index if w.end_time.normalize() <= as_of]
    last = done[-1]
    elig = [w for w in done if w.end_time.normalize() + pd.Timedelta(days=15) <= as_of]
    coh = elig[-1]
    med = W.loc[done[-13:-1], "capture_ratio"].median()
    r, c = W.loc[last], W.loc[coh]

    matched = M["paid_rs"].sum(); all_cap = cap["amount_paise"].sum() / 100
    fees, tax, settled = [S[x].sum() / 100 for x in ("fees_paise", "tax_paise", "amount_paise")]
    tie1 = abs(matched - all_cap) < 1
    tie2 = abs(all_cap - fees - tax - unsettled_rs - settled) < 1

    rows = [
      (1, f"capture ratio, week {last}",    f"{r['capture_ratio']:.3f}", f"median {med:.3f}", r["capture_ratio"] < THRESH["ratio_min"] * med),
      (2, f"unpaid at day 15, cohort {coh}", f"{c['unpaid15']:.3%}",      "about 2.9%",        c["unpaid15"] > THRESH["unpaid15_max"]),
      (3, f"refs needing fix, week {last}", f"{r['fix_rate']:.2%}",       "about 1.0%",        not (THRESH["fix_band"][0] <= r["fix_rate"] <= THRESH["fix_band"][1])),
      (4, f"payments with no invoice, week {last}", int(r["unmatched"]),  "0",                 r["unmatched"] > 0),
      (5, f"extra-capture rate, week {last}", f"{r['extra_rate']:.2%}",   "about 0.4%",        r["extra_rate"] > THRESH["extra_max"]),
      (6, f"bad settlements, week {last}",  int(r["bad_settlements"]) if r["bad_settlements"] == r["bad_settlements"] else 0, "0", (r["bad_settlements"] or 0) > 0),
      (7, "payments older than 36h, no settlement row", stuck,            "0",                 stuck > 0),
      (8, "rupee ties hold",                f"{tie1 and tie2}",           "True",              not (tie1 and tie2)),
    ]
    out = pd.DataFrame(rows, columns=["#", "check", "value", "normal", "alert"])
    out["status"] = np.where(out["alert"], "ALERT", "ok")
    return out.drop(columns="alert"), W, done, as_of

result, W, done, as_of = weekly_check(M, G, S)
print(result.to_string(index=False))

# backtest: would these rules have fired in the past?
Wd = W.loc[done].copy()
Wd["ratio_alert"] = Wd["capture_ratio"] < THRESH["ratio_min"] * Wd["capture_ratio"].rolling(12).median().shift(1)
ok_age = np.array([w.end_time.normalize() + pd.Timedelta(days=15) <= as_of for w in Wd.index])
Wd["unpaid15_alert"] = (Wd["unpaid15"] > THRESH["unpaid15_max"]) & ok_age
print(Wd[Wd[["ratio_alert", "unpaid15_alert"]].any(axis=1)][["invoices", "captures", "capture_ratio", "unpaid15"]])
print(Wd[["capture_ratio", "unpaid15", "fix_rate", "extra_rate"]].describe().round(4))   # use this to recalibrate
```
